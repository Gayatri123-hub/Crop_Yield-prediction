```python
categorical_features = [
    "Crop",
    "Season",
    "State"
]

numerical_features = [
    "Year",
    "Area (hectares)",
    "Fertilizer (kg)",
    "Pesticide (kg)",
    "Avg Temperature (C)",
    "Total Rainfall (mm)",
    "Avg Humidity (%)",
    "Nitrogen (N)",
    "Phosphorus (P)",
    "Potassium (K)",
    "Soil pH"
]

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
# Sort the dataset by year
df = df.sort_values("Year")

# Training data: years before 2018
train_df = df[df["Year"] < 2018]

# Testing data: 2018 onwards
test_df = df[df["Year"] >= 2018]

print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)

print("\nTraining years:")
print(train_df["Year"].min(), "to", train_df["Year"].max())

print("\nTesting years:")
print(test_df["Year"].min(), "to", test_df["Year"].max())
```

```python
target = "Yield (tonnes/hectare)"

# Separate input features and target
X_train = train_df.drop(target, axis=1)
y_train = train_df[target]

X_test = test_df.drop(target, axis=1)
y_test = test_df[target]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["Crop", "Season", "State"]

numerical_features = [
    "Year",
    "Area (hectares)",
    "Fertilizer (kg)",
    "Pesticide (kg)",
    "Avg Temperature (C)",
    "Total Rainfall (mm)",
    "Avg Humidity (%)",
    "Nitrogen (N)",
    "Phosphorus (P)",
    "Potassium (K)",
    "Soil pH"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("numerical", "passthrough", numerical_features)
    ]
)

print("Preprocessor created successfully")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

print("Linear Regression pipeline created successfully")
```

```python
# Train the Linear Regression model
linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully")
```

```python
# Predict yield for the test data
y_pred = linear_model.predict(X_test)

print("Predictions generated successfully")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
from sklearn.model_selection import train_test_split

train_df = df[df["Year"] <= 2017]
test_df = df[df["Year"] >= 2018]

X_train = train_df.drop("Yield (tonnes/hectare)", axis=1)
y_train = train_df["Yield (tonnes/hectare)"]

X_test = test_df.drop("Yield (tonnes/hectare)", axis=1)
y_test = test_df["Yield (tonnes/hectare)"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
 from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Categorical columns
categorical_features = [
    "Crop",
    "Season",
    "State"
]

# Numerical columns
numerical_features = [
    "Year",
    "Area (hectares)",
    "Fertilizer (kg)",
    "Pesticide (kg)",
    "Avg Temperature (C)",
    "Total Rainfall (mm)",
    "Avg Humidity (%)",
    "Nitrogen (N)",
    "Phosphorus (P)",
    "Potassium (K)",
    "Soil pH"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessor created successfully")
```

```python
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

xgb_model = XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    objective="reg:squarederror"
)

xgb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", xgb_model)
    ]
)

print("XGBoost pipeline created successfully")
```

```python
xgb_pipeline.fit(X_train, y_train)

print("XGBoost model trained successfully")
```

```python
y_pred_xgb = xgb_pipeline.predict(X_test)

print("XGBoost predictions completed successfully")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = r2_score(y_test, y_pred_xgb)

print("XGBoost Performance")
print("MAE :", mae_xgb)
print("RMSE:", rmse_xgb)
print("R²  :", r2_xgb)
```

```python
import pandas as pd

def recommend_crops(year, season, state, area, fertilizer, pesticide,
                    temperature, rainfall, humidity,
                    nitrogen, phosphorus, potassium, soil_ph):

    # Step 1: Find crops available for selected State and Season
    available_crops = df[
        (df["State"] == state) &
        (df["Season"] == season)
    ]["Crop"].unique()

    recommendations = []

    # Step 2: Predict yield for each available crop
    for crop in available_crops:

        input_data = pd.DataFrame([{
            "Crop": crop,
            "Year": year,
            "Season": season,
            "State": state,
            "Area (hectares)": area,
            "Fertilizer (kg)": fertilizer,
            "Pesticide (kg)": pesticide,
            "Avg Temperature (C)": temperature,
            "Total Rainfall (mm)": rainfall,
            "Avg Humidity (%)": humidity,
            "Nitrogen (N)": nitrogen,
            "Phosphorus (P)": phosphorus,
            "Potassium (K)": potassium,
            "Soil pH": soil_ph
        }])

        predicted_yield = xgb_pipeline.predict(input_data)[0]

        recommendations.append({
            "Crop": crop,
            "Predicted Yield (tonnes/hectare)": predicted_yield
        })

    # Step 3: Convert to DataFrame
    result = pd.DataFrame(recommendations)

    # Step 4: Sort crops by predicted yield
    result = result.sort_values(
        by="Predicted Yield (tonnes/hectare)",
        ascending=False
    )

    # Step 5: Add ranking
    result.insert(
        0,
        "Rank",
        range(1, len(result) + 1)
    )

    return result.reset_index(drop=True)
```

```python
def what_if_analysis(
    crop,
    year,
    season,
    state,
    area,
    fertilizer,
    pesticide,
    temperature,
    rainfall,
    humidity,
    nitrogen,
    phosphorus,
    potassium,
    soil_ph,
    variable,
    values
):

    results = []

    for value in values:

        current_fertilizer = fertilizer
        current_pesticide = pesticide
        current_temperature = temperature
        current_rainfall = rainfall
        current_humidity = humidity
        current_nitrogen = nitrogen
        current_phosphorus = phosphorus
        current_potassium = potassium
        current_soil_ph = soil_ph

        if variable == "Fertilizer":
            current_fertilizer = value

        elif variable == "Pesticide":
            current_pesticide = value

        elif variable == "Temperature":
            current_temperature = value

        elif variable == "Rainfall":
            current_rainfall = value

        elif variable == "Humidity":
            current_humidity = value

        elif variable == "Nitrogen":
            current_nitrogen = value

        elif variable == "Phosphorus":
            current_phosphorus = value

        elif variable == "Potassium":
            current_potassium = value

        elif variable == "Soil pH":
            current_soil_ph = value

        predicted_yield = what_if_yield(
            crop=crop,
            year=year,
            season=season,
            state=state,
            area=area,
            fertilizer=current_fertilizer,
            pesticide=current_pesticide,
            temperature=current_temperature,
            rainfall=current_rainfall,
            humidity=current_humidity,
            nitrogen=current_nitrogen,
            phosphorus=current_phosphorus,
            potassium=current_potassium,
            soil_ph=current_soil_ph
        )

        results.append({
            variable: value,
            "Predicted Yield (tonnes/hectare)": predicted_yield
        })

    return pd.DataFrame(results)
```

```python
def run_what_if(
    crop,
    year,
    season,
    state,
    area,
    fertilizer,
    pesticide,
    temperature,
    rainfall,
    humidity,
    nitrogen,
    phosphorus,
    potassium,
    soil_ph,
    variable,
    values
):

    results = []

    for value in values:

        # Keep all original values
        current_values = {
            "fertilizer": fertilizer,
            "pesticide": pesticide,
            "temperature": temperature,
            "rainfall": rainfall,
            "humidity": humidity,
            "nitrogen": nitrogen,
            "phosphorus": phosphorus,
            "potassium": potassium,
            "soil_ph": soil_ph
        }

        # Change only the selected variable
        if variable == "Fertilizer":
            current_values["fertilizer"] = value

        elif variable == "Pesticide":
            current_values["pesticide"] = value

        elif variable == "Temperature":
            current_values["temperature"] = value

        elif variable == "Rainfall":
            current_values["rainfall"] = value

        elif variable == "Humidity":
            current_values["humidity"] = value

        elif variable == "Nitrogen":
            current_values["nitrogen"] = value

        elif variable == "Phosphorus":
            current_values["phosphorus"] = value

        elif variable == "Potassium":
            current_values["potassium"] = value

        elif variable == "Soil pH":
            current_values["soil_ph"] = value

        # Predict yield
        predicted_yield = what_if_yield(
            crop=crop,
            year=year,
            season=season,
            state=state,
            area=area,
            fertilizer=current_values["fertilizer"],
            pesticide=current_values["pesticide"],
            temperature=current_values["temperature"],
            rainfall=current_values["rainfall"],
            humidity=current_values["humidity"],
            nitrogen=current_values["nitrogen"],
            phosphorus=current_values["phosphorus"],
            potassium=current_values["potassium"],
            soil_ph=current_values["soil_ph"]
        )

        results.append({
            variable: value,
            "Predicted Yield (tonnes/hectare)": predicted_yield
        })

    return pd.DataFrame(results)
```

```python
# Split data by year
train_df = df[df["Year"] <= 2017].copy()
test_df = df[df["Year"] >= 2018].copy()

# Separate input features and target
X_train = train_df.drop("Yield (tonnes/hectare)", axis=1)
y_train = train_df["Yield (tonnes/hectare)"]

X_test = test_df.drop("Yield (tonnes/hectare)", axis=1)
y_test = test_df["Yield (tonnes/hectare)"]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Data split completed successfully!")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_features = ["Crop", "Season", "State"]

numerical_features = [
    "Year",
    "Area (hectares)",
    "Fertilizer (kg)",
    "Pesticide (kg)",
    "Avg Temperature (C)",
    "Total Rainfall (mm)",
    "Avg Humidity (%)",
    "Nitrogen (N)",
    "Phosphorus (P)",
    "Potassium (K)",
    "Soil pH"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessor created successfully!")
```

```python
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

xgb_model = XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    objective="reg:squarederror"
)

xgb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", xgb_model)
    ]
)

print("Training started...")

xgb_pipeline.fit(X_train, y_train)

print("XGBoost model trained successfully!")
```
